```python
import os
import json
import joblib
import pandas as pd
import matplotlib.pyplot as plt
import shap

df = pd.read_csv("../data/processed_loans.csv")
X = df.drop(columns=["target"])
y = df["target"]

xgb = joblib.load("../models/xgb_model.pkl")

os.makedirs("../models", exist_ok=True)
```

```python
explainer = shap.TreeExplainer(xgb)

X_small = X.sample(n=min(400, len(X)), random_state=42)
shap_values = explainer(X_small)

shap.summary_plot(shap_values, X_small, show=False)
plt.tight_layout()
plt.savefig("../models/shap_summary.png", dpi=200, bbox_inches="tight")
plt.close()

print("Saved: ../models/shap_summary.png")
```

```python
one = X_small.iloc[[0]]
one_sv = explainer(one)

shap.plots.waterfall(one_sv[0], show=False)
plt.tight_layout()
plt.savefig("../models/shap_waterfall.png", dpi=200, bbox_inches="tight")
plt.close()

print("Saved: ../models/shap_waterfall.png")
```

```python
import json, joblib, pandas as pd, shap, matplotlib.pyplot as plt

xgb = joblib.load("../models/xgb_model.pkl")
cols = json.load(open("../models/feature_columns.json"))

df = pd.read_csv("../data/processed_loans.csv")
X = df.drop(columns=["target"])[cols]

explainer = shap.TreeExplainer(xgb)

X_bg = X.sample(n=min(400, len(X)), random_state=42)
sv_bg = explainer(X_bg)

shap.summary_plot(sv_bg, X_bg, show=False)
plt.tight_layout()
plt.savefig("../models/shap_summary_updated.png", dpi=200, bbox_inches="tight")
plt.close()

print("Saved:", "../models/shap_summary_updated.png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import shap

income_annum = 2000000
loan_amount = 12000000
loan_term = 10
cibil_score = 651
no_of_dependents = 6
education = "Not Graduate"
self_employed = "Yes"
asset_total = 100000

emi_income_ratio = (loan_amount / loan_term) / income_annum

if cibil_score < 550:
    cibil_band = 0
elif cibil_score < 650:
    cibil_band = 1
elif cibil_score < 750:
    cibil_band = 2
else:
    cibil_band = 3

row = {
    "income_annum": income_annum,
    "emi_income_ratio": emi_income_ratio,
    "no_of_dependents": no_of_dependents,
    "is_not_graduate": 1 if education == "Not Graduate" else 0,
    "is_self_employed": 1 if self_employed == "Yes" else 0,
    "asset_total": asset_total,
    "cibil_band": cibil_band,
}

X_one = pd.DataFrame([row])[cols]
sv_one = explainer(X_one)

shap.plots.waterfall(sv_one[0], show=False)
plt.tight_layout()
plt.savefig("../models/shap_waterfall_case_streamlit.png", dpi=200, bbox_inches="tight")
plt.close()

print("cibil_band used:", cibil_band)
print("Saved:", "../models/shap_waterfall_case_streamlit.png")
```
